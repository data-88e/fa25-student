# Sympy for Symbolic Mathematics

```python
import sympy as sp

# Use pretty printing for math-like output
sp.init_printing()

# Declare q as a symbolic variable
q = sp.Symbol('q')
```

```python
# Total cost function
TC = 50 + 10*q + q**2

# Price function
P = 50 - 2*q

# Revenue is price × quantity
R = P * q

# Profit is revenue - cost
π = R - TC
```

```python
# Show the profit function
π
```

```python
# Take the derivative of profit with respect to q
dπ_dq = sp.diff(π, q)
dπ_dq
```

```python
# Set first derivative = 0 and solve
q_star = sp.solve(dπ_dq, q)
q_star
```

```python
# Second derivative
second_derivative = sp.diff(dπ_dq, q)
second_derivative.subs(q, q_star[0])
```

```python
# Substitute optimal q into functions
R_star = R.subs(q, q_star[0])
TC_star = TC.subs(q, q_star[0])
π_star = π.subs(q, q_star[0])

R_star, TC_star, π_star
```

```python
graph = sp.plot(π, (q, 0, 10), show=False)
graph.title = "Profit as a function of quantity"
graph.xlabel = "Quantity"
graph.ylabel = "Profit"
graph.show()
```

```python
# Plot profit, revenue, and cost on the same axes
p1 = sp.plot(π, (q, 0, 10), line_color='blue', show=False, label='Profit')
p2 = sp.plot(R, (q, 0, 10), line_color='green', show=False, label='Revenue')
p3 = sp.plot(TC, (q, 0, 10), line_color='red', show=False, label='Total Cost')
p1.extend(p2)
p1.extend(p3)
p1.title = "Profit, Revenue, and Cost"
p1.xlabel = "Quantity"
p1.ylabel = "Dollars"
p1.legend = True
p1.show()
```
