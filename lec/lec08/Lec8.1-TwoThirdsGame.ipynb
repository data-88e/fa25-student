# Guess 2/3 of the Average Game

## How the Game Works

This is a classic game theory exercise that demonstrates strategic thinking and convergent reasoning.

**Rules:**
1. Each player chooses a number between 0 and 100
2. The "target" is calculated as **2/3 of the average** of all guesses
3. The winner is the player whose guess is **closest to the target**


Here is the wiki page for more details: [Guess 2/3 of the Average](https://en.wikipedia.org/wiki/Guess_2/3_of_the_Average)


## This Simulation

We simulate 3 rounds where players iteratively adjust their guesses based on the previous round's target, to see if there is a trend toward convergence.

```python
from datascience import *
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

## Round 1

```python
guesses1 = [xx, xx, xx, xx, xx, xx, xx]
```

```python
#ignore these other guesses!
#guesses1 = [37,45,20]
# get 7 random numbers between 0 and 100 using numpy
#guesses1 = list(np.random.randint(0, 101, 7))
```

```python
avg1 = np.mean(guesses1)
target1 = (2/3) * avg1

```

```python
print('Round 1 average:', round(avg1,2))
print('Round 1 target:', round(target1,2))
```

```python
# Find the closest guess to the target
differences = np.abs(guesses1 - target1)
closest_index = np.argmin(differences)
closest_guess = guesses1[closest_index]

print('Winner: Person', closest_index + 1)
print('Their guess:', closest_guess)
print('Distance from target:', round(differences[closest_index], 2))
```

## Game Theory Insights

### this is what the economic theory book says

- **First-level thinking**: "The average will be 50, so I'll guess 2/3 × 50 = 33"
- **Second-level thinking**: "Others will guess 33, so I'll guess 2/3 × 33 = 22"
- **Nash Equilibrium**: With rational players, this converges to **0** (the only stable solution)
